# Week 1: Basic Descriptive Statistics & Exploratory Data Analysis

```python
import statistics

# Input data
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

#  Total
count = len(ages)
print(f"a. Survey size: {count} respondents")

# Total ages
sumAges = sum(ages)
print(f"b. Sum of ages: {sumAges}")

# Rounded to one decimal place
meanAges = sumAges / count
print(f"c. Mean age: {meanAges:.1f}")

# Min and Max ages
minAges = min(ages)
maxAges = max(ages)
print(f"d. Min: {minAges} | Max: {maxAges}")

# e. Median calculation
median_age = statistics.median(ages)
print(f"e. Median age: {median_age} years")
print(f"e. It tells you the distribution is skewed by extreme values (outliers")
```
