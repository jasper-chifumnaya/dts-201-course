# Week 1: Basic Descriptive Statistics

```python
import statistics

# Input data
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

# Total
count = len(ages)
print(f"a. Number of responses: {count}")

# Sum of ages
sumAges = sum(ages)
print(f"b. Sum of ages: {sumAges}")

# Mean age 
meanAges = sumAges / count
print(f"c. Mean age: {meanAges:.1f}")

# Min and Max ages
minAges = min(ages)
maxAges = max(ages)
print(f"d. Min: {minAges} | Max: {maxAges}")

# Median
median_age = statistics.median(ages)
print(f"e. Median age: {median_age}")
```
