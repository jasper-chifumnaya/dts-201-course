# Week 1: Basic Descriptive Statistics

```python
import statistics

# Input data
ages = [19, 22, 21, 35, 28, 24, 31, 19, 27, 40]

# Total
count = len(ages)
print(f"a. Number of responses: {count}")

# Sum of ages
sumAges = sum(ages)
print(f"b. Sum of ages: {sumAges}")

# Mean age 
meanAges = sumAges / count
print(f"c. Mean age: {meanAges:.1f}")

# Min and Max ages
minAges = min(ages)
maxAges = max(ages)
print(f"d. Min: {minAges} | Max: {maxAges}")

# Median
median_age = statistics.median(ages)
print(f"e. Median age: {median_age}")
```

# Week 3: Data Collection Smapling

```python
import pandas as pd

# Load Excel file
df_messy = pd.read_excel("50_startups_messy.xlsx")

# 1. Shape
print(f"Shape: {df_messy.shape}")
print(f"Rows: {df_messy.shape[0]}, Columns: {df_messy.shape[1]}")

# 2. Missing counts per column
print("\n--- Missing Values Count ---")
print(df_messy.isnull().sum())

# 3. Duplicate rows count
num_duplicates = df_messy.duplicated().sum()
print(f"\nDuplicate rows count: {num_duplicates}")
```

```python
# Make a copy
df_clean = df_messy.copy()

# a. Drop duplicate rows
df_clean = df_clean.drop_duplicates()

# 1. Clean all column names (removes hidden leading/trailing spaces and standardizes)
df_clean.columns = df_clean.columns.str.strip()

# Check to confirm 'State' is now present
print("Cleaned Columns:", df_clean.columns.tolist())

# b. Standardise 'State' (strip leading/trailing whitespaces, capitalize appropriately)
df_clean['State'] = df_clean['State'].astype(str).str.strip().str.title()

# Fix typos if present (e.g. 'Cal', 'Ny', 'Fl')
state_mapping = {
    'Ny': 'New York',
    'Nyc': 'New York',
    'Newyork': 'New York',
    'Cal': 'California',
    'Calif': 'California',
    'Fl': 'Florida',
    'Fla': 'Florida'
}
df_clean['State'] = df_clean['State'].replace(state_mapping)

# c. Coerce 'Marketing Spend' to numeric (turn invalid text/strings into NaN)
df_clean['Marketing Spend'] = pd.to_numeric(df_clean['Marketing Spend'], errors='coerce')

# Verify results
print("Cleaned DataFrame info:")
df_clean.info()
print("\nUnique States:", df_clean['State'].unique())
```

```python
print(df_messy.columns.tolist())
```

```python
# 1. Simple Random Sample (n = 20, random_state = 7)
simple_sample = df_clean.sample(n=20, random_state=7)

# 2. Stratified 20% Sample by 'State'
stratified_sample = df_clean.groupby('State', group_keys=False).apply(
    lambda x: x.sample(frac=0.20, random_state=7)
)

print(f"Simple Random Sample size: {len(simple_sample)}")
print(f"Stratified Sample size: {len(stratified_sample)}")
print("\nStratified Sample State Breakdown:\n", stratified_sample['State'].value_counts())
```

```python
# Calculate means for R&D Spend and Marketing Spend
features = ['R&D Spend', 'Marketing Spend']

pop_means = df_clean[features].mean()
simple_means = simple_sample[features].mean()
stratified_means = stratified_sample[features].mean()

# Build comparison table
comparison_df = pd.DataFrame({
    'Population Mean': pop_means,
    'Simple Random Mean (n=20)': simple_means,
    'Stratified Mean (20%)': stratified_means,
    'Diff (Simple - Pop)': simple_means - pop_means,
    'Diff (Stratified - Pop)': stratified_means - pop_means
})

comparison_df
```
